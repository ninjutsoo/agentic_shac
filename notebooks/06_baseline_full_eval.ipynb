# 06_baseline_full_eval

Evaluate the baseline model on dev/test splits, save predictions, and inspect metrics (Accuracy, FPR, confusion matrices).


```python
# Setup
import json
from pathlib import Path

# Resolve project root robustly in notebooks (no __file__)
CWD = Path.cwd()
if (CWD / 'configs').exists() and (CWD / 'src').exists():
    PROJECT_ROOT = CWD
elif CWD.name == 'notebooks' and (CWD.parent / 'configs').exists():
    PROJECT_ROOT = CWD.parent
else:
    cur = CWD
    PROJECT_ROOT = CWD
    while cur != cur.parent:
        if (cur / 'configs').exists() and (cur / 'src').exists():
            PROJECT_ROOT = cur
            break
        cur = cur.parent

CONFIG = PROJECT_ROOT / 'configs' / 'baseline.yaml'
DATA_DIR = PROJECT_ROOT / 'data' / 'processed'
OUT_DIR = PROJECT_ROOT / 'experiments' / 'baseline'

print('Project root:', PROJECT_ROOT)
print('Config:', CONFIG)
print('Data dir:', DATA_DIR)
print('Out dir:', OUT_DIR)
```

```python
# Run baseline on dev and test (idempotent)
import subprocess, sys

splits = ['dev', 'test']
for split in splits:
    cmd = [sys.executable, '-m', 'src.evaluation.run_baseline', '--split', split, '--config', str(CONFIG), '--data_dir', str(DATA_DIR), '--out_dir', str(OUT_DIR)]
    print('\nRunning:', ' '.join(cmd))
    completed = subprocess.run(cmd, cwd=PROJECT_ROOT, check=True)
print('\nDone.')
```

```python
# Load latest metrics for each split (independently)
import glob
import json

for split in ['dev', 'test']:
    pattern = str(OUT_DIR / '*' / f'metrics_{split}.json')
    candidates = sorted(glob.glob(pattern))
    if not candidates:
        print(f"No metrics found for split '{split}'.")
        continue
    metrics_path = Path(candidates[-1])
    print(f"Latest metrics for {split}: {metrics_path}")
    with open(metrics_path, 'r') as f:
        metrics = json.load(f)
    print(f"\n=== {split.upper()} ===")
    print('Accuracy:', metrics['accuracy'])
    print('FPR:', metrics['fpr'])
    print('Labels:', metrics['labels'])
    print('True dist:', metrics['true_distribution'])
    print('Pred dist:', metrics['pred_distribution'])
```
